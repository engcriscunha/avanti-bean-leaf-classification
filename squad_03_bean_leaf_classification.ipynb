```python
def verify_image_paths(df, dataset_base_path):
    missing_images = []
    for index, row in df.iterrows():
        image_relative_path = row['image:FILE']
        full_image_path = os.path.join(dataset_base_path, image_relative_path)
        if not os.path.exists(full_image_path):
            missing_images.append(image_relative_path)
    return missing_images

print("Verificando imagens do dataset de treinamento...")
missing_train_images = verify_image_paths(train_df, dataset_path)

if missing_train_images:
     print(f"ATENÇÃO: {len(missing_train_images)} imagens ausentes no dataset de treinamento:")
     for img in missing_train_images[:10]: # Print only first 10 for brevity
         print(img)

else:
     print("Todas as imagens do dataset de treinamento foram encontradas.")

print("\nVerificando imagens do dataset de validação...")
missing_val_images = verify_image_paths(val_df, dataset_path)

if missing_val_images:
    print(f"ATENÇÃO: {len(missing_val_images)} imagens ausentes no dataset de validação:")
    for img in missing_val_images[:10]: # Print only first 10 for brevity
        print(img)
else:
    print("Todas as imagens do dataset de validação foram encontradas.")
```
